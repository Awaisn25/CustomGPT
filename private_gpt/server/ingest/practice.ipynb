```python
from PIL import Image
import ollama

# Load image (adjust the path as needed)
image_path1 = '/home/awaisn/Pictures/Screenshots/Screenshot from 2025-12-02 12-27-51.png'
with open(image_path1, 'rb') as img_file:
    image_bytes = img_file.read()

# Prepare prompt
prompt = "Give a count of how many VDE1, VDE2 and RFB1 do you see in this image."

# Call ollama qwen3-vl model with an image
response = ollama.chat(
    model='qwen3-vl:2b',
    messages=[
        {
            'role': 'user',
            'content': prompt,
            'images': [image_path1]
        }
    ]
)

print(response['message']['content'])
```
